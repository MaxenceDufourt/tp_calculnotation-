```python
import matplotlib.pyplot  
import numpy as np 

liste_note = [12,14,10,13,11,12,12,14,14,16,10.5,15,13,14,15,16,17,18,19,14,9]
liste_coeffUe1 = [ 10,10,7,7,0,5,0,6,0,5,4,2,5,5,0,20,20,0,0,0,7]
liste_coeffUe2 = [ 4,0,2,8,6,0,0,0,0,5,5,2,9,9,3,0,0,29,0,0,7 ]
liste_coeffUe3 = [ 4,0,2,0,0,5,15,6,4,5,5,2,0,0,3,0,0,0,20,20,7]
totalnoteUe1 = 0
totalnoteUe2 = 0
totalnoteUe3 = 0
totalcoeffUe1 = 0
totalcoeffUe2 = 0
totalcoeffUe3 = 0 
for i in range (21) : # i voyage sur la longueurs de 0 à 21 et multiplie chaque valeur de note avec un coeff 
    totalnoteUe1 += liste_note[i] * liste_coeffUe1[i]  # calcul du total des notes x le coeff de l'UE 1
    totalcoeffUe1 += liste_coeffUe1[i]  # fait la sommes des coeffs
    moyUE1 = totalnoteUe1 / totalcoeffUe1 # fait la moyenne du résultats totale des notes*coeff UE3  divisé par le sommes totale des coeffs

for i in range (21) :  # i voyage sur la longueurs de 0 à 21 et multiplie chaque valeur de note avec un coeff 
    totalnoteUe2 += liste_note[i] * liste_coeffUe2[i]  # calcul du total des notes x le coeff de l'UE 2
    totalcoeffUe2 += liste_coeffUe2[i] # fait la sommes des coeffs
    moyUE2 = totalnoteUe2 / totalcoeffUe2 # fait la moyenne du résultats totale des notes*coeff UE2  divisé par le sommes totale des coeffs

for i in range (21) :   # i voyage sur la longueurs de 0 à 21 et multiplie chaque valeur de note avec un coeff 
    totalnoteUe3 += liste_note[i] * liste_coeffUe3[i] # calcul du total des notes x le coeff de l'UE 3 
    totalcoeffUe3 += liste_coeffUe3[i] # fait la sommes des coeffs 
    moyUE3 = totalnoteUe3 / totalcoeffUe3   # fait la moyenne du résultats totale des notes*coeff UE3  divisé par le sommes totale des coeffs
print("Voici t'a moyenne de UE1 " ,moyUE1)
print("Voici t'a moyenne de UE2 " ,moyUE2)
print("Voici t'a moyenne de UE3 " ,moyUE3)


```
